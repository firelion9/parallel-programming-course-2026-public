```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
mapping = {
    "V0": "**Этап 0:** Однопоточный baseline",
    "V1a": "**Этап 1:** Пустой лок (без записи)",
    "V1": "**Этап 1:** Единый общий лок",
    "V2": "**Этап 2:** Шардированный лок (16 групп)",
    "V3": "**Этап 3:** Thread-Local состояние (локальные данные)",
    "V4": "**Этап 4:** Двойная буферизация",
}

def process(path: str):
    data = pd.read_csv(path)
    lines = list(data.groupby(data["Ver"]))
    for i in range(len(lines)):
        plt.plot(lines[i][1]["thr"], lines[i][1]["Gops/s"], label=lines[i][0])
        plt.legend()
        plt.xlabel('n_processes')
        plt.ylabel('Gops/s')
    lines = {k: v for k, v in lines}
    for key, title in mapping.items():
        print("|" + title + "|" + "|".join(lines[key]['Gops/s'] * 1000) + "|")
```

```python
process("runs/run01.csv")
```

```python
process("runs/run02.csv")
```
